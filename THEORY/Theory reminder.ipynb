# Theory Reminder

This section summarizes the main equations and concepts used in the flow and transport models. It is meant as a reference, not as a complete derivation.

---

## Flow

### 01f - Main variables and properties

| Symbol | Meaning | Typical unit |
| :--- | :--- | :--- |
| $Q$ | Volumetric discharge | $\mathrm{m^3/s}$ or L/s |
| $\mathbf q$ | Specific discharge (Darcy flux) | $\mathrm{m/s}$ |
| $K$ | Hydraulic conductivity | $\mathrm{m/s}$ |
| $\phi$ | Porosity | – |
| $\theta_w$ | Volumetric water content | – |
| $S_w$ | Water saturation | – |
| $h$ | Hydraulic head | m |
| $T$ | Transmissivity | $\mathrm{m^2/s}$ |

> **Note:** $Q$ and $\mathbf q$ are distinct quantities. $Q$ is a total flow rate, while $\mathbf q$ is flow per unit cross-sectional area:
>
> $$Q = \mathbf q \cdot \mathbf A.$$

### 02f - Hydraulic head

The hydraulic head is the sum of elevation head and pressure head:

$$h = z + \frac{p}{\rho g}$$

where $z$ is elevation, $p$ is pressure, $\rho$ is water density, and $g$ is gravitational acceleration.

### 03f - Darcy's law

Darcy's law relates the specific discharge to the hydraulic gradient:

$$\mathbf q = -\mathbf K \nabla h$$

For a homogeneous isotropic aquifer:

$$\mathbf q = -K\nabla h.$$

For a simple 1D case:

$$q = -K\frac{\Delta h}{L}.$$

The total discharge through a cross-sectional area $A$ is:

$$Q = qA.$$

The minus sign indicates that groundwater flows from high hydraulic head towards low hydraulic head.

### 04f - Equivalent hydraulic conductivity

For layers **parallel to flow** (e.g. horizontal flow through horizontal layers):

$$K_{\mathrm{eq}} = \frac{\sum_i K_i d_i}{\sum_i d_i}$$

For layers **perpendicular to flow** (e.g. vertical flow through horizontal layers):

$$\frac{1}{K_{\mathrm{eq}}} = \frac{\sum_i d_i/K_i}{\sum_i d_i}$$

where $d_i$ is the thickness of layer $i$.

### 05f - Transmissivity

For a confined aquifer:

$$T = Kb$$

where $b$ is the saturated aquifer thickness.

For an unconfined aquifer:

$$T = K(h-z_{\mathrm{bottom}})$$

where $h-z_{\mathrm{bottom}}$ is the saturated thickness.

### 06f - Mass balance and storage

For constant water density, the groundwater mass balance can be written as:

$$S_0\frac{\partial h}{\partial t} = -\nabla\cdot\mathbf q + W$$

Using Darcy's law:

$$S_0\frac{\partial h}{\partial t} = \nabla\cdot(K\nabla h)+W$$

where $W$ represents sources or sinks of water.

#### Storage parameters

| Symbol | Meaning | Role | Unit |
| :--- | :--- | :--- | :--- |
| $S_0$ | Specific storage | Compressibility of aquifer and water | $\mathrm{m^{-1}}$ |
| $S$ | Storativity | Storage response of the aquifer | - |
| $\phi_d$ | Drainable porosity / specific yield | Gravity drainage in an unconfined aquifer | - |

Specific storage is the volume of water released from storage per unit aquifer volume and per unit decline in hydraulic head:

$$S_0 = \rho g\left(\alpha_{\mathrm{aq}}+\phi\beta_w\right)$$

where:

- $\alpha_{\mathrm{aq}}$ is the aquifer compressibility [$\mathrm{Pa^{-1}}$];
- $\beta_w$ is the water compressibility [$\mathrm{Pa^{-1}}$].

For a confined aquifer:

$$S=S_0b$$

For an unconfined aquifer, storage is often approximated using the drainable porosity (specific yield):

$$S\approx\phi_d$$

### 07f - Sources and sinks

$W$ represents water added to or removed from the aquifer, for example:

- recharge;
- lateral inflow or outflow;
- pumping wells;
- injection wells;
- river leakage.

### 08f - Boundary conditions

#### 1. Specified head — Dirichlet boundary

For example, a river boundary:

$$h(x,y,t)=h_{\mathrm{riv}}(x,y,t)$$

#### 2. Specified flux — Neumann boundary

For example, recharge:

$$q_n=N$$

where $N$ is the prescribed flux normal to the boundary.

#### 3. Impermeable boundary

No flow crosses the boundary:

$$\mathbf q\cdot\mathbf n=0$$

For constant isotropic $K$, this is equivalent to:

$$\frac{\partial h}{\partial n}=0$$

### 09f - Simple analytical solutions

#### 1 - Steady state

At steady state, hydraulic head does not change with time:

$$\frac{\partial h}{\partial t}=0$$

#### 2 - Confined 1D flow

For a homogeneous aquifer without sources or sinks:

$$\frac{d^2h}{dx^2}=0.$$

The hydraulic head varies linearly:

$$h(x)=h_0+\frac{h_L-h_0}{L}x$$

The discharge through an aquifer of width $B_y$ is:

$$Q=-TB_y\frac{h_L-h_0}{L}$$

#### 3 - Unconfined 1D flow — Dupuit approximation

For steady flow with uniform recharge $N$:

$$h(x)^2=h_0^2-(h_0^2-h_L^2)\frac{x}{L}+\frac{N}{K}x(x-L)$$

#### 4 - Confined 2D radial flow

For a well at the origin:

$$h(r)=h(R)+\frac{Q_w}{2\pi T}\ln\left(\frac{R}{r}\right)$$

where:

- $Q_w<0$ for pumping (extraction);
- $Q_w>0$ for injection;
- $R$ is a reference radius where $h=h(R)$.

For pumping, $Q_w<0$, so:

$$h(r)<h(R)$$

as expected.

An approximate transient radius of influence is:

$$R(t)\approx1.5\sqrt{\frac{T}{S}t}$$

#### 5 - Unconfined 2D radial flow

Under the Dupuit approximation:

$$h(r)^2=h(R)^2+\frac{Q_w}{\pi K}\ln\left(\frac{R}{r}\right)$$

### 03t - Transport equation

The general advection-dispersion-reaction equation is:

$$\frac{\partial(\phi_e R c)}{\partial t}=-\nabla\cdot(\phi_e\mathbf u c)+\nabla\cdot\left[\phi_e(\mathbf D_m+\mathbf D)\nabla c\right]-\lambda\phi_eRc+P$$

For a conservative solute:

$$R=1,\qquad\lambda=0.$$

For constant $\phi_e$, $R$, $\mathbf u$, and $\mathbf D$:

$$\frac{\partial c}{\partial t}=-\mathbf u\cdot\nabla c+\nabla\cdot\left[(\mathbf D_m+\mathbf D)\nabla c\right]-\lambda c$$

### 04t - Advection

The seepage velocity is related to the Darcy flux by:

$$\mathbf u=\frac{\mathbf q}{\phi_e}$$

The advective solute mass flux is:

$$\mathbf j_a=\phi_e\mathbf u c=\mathbf q c$$

Advection therefore transports solute with the groundwater.

### 05t - Molecular diffusion

Molecular diffusion is driven by concentration gradients and follows Fick's law:

$$\mathbf j_m=-\phi_e\mathbf D_m\nabla c$$

For isotropic molecular diffusion:

$$\mathbf D_m=D_m^*\mathbf I.$$

A typical molecular diffusivity in water is:

$$D_m^*\approx10^{-9}\ \mathrm{m^2/s}$$

If the model uses days as the time unit:

$$D_m^*\approx8.64\times10^{-5}\ \mathrm{m^2/d}$$

### 06t - Mechanical dispersion

Mechanical dispersion results from variations in groundwater velocity at the pore scale.

It is described using:

- $\alpha_L$: longitudinal dispersivity [m];
- $\alpha_T$: transverse dispersivity [m].

The dispersion tensor is:

$$\mathbf D=\alpha_T|\mathbf u|\mathbf I+(\alpha_L-\alpha_T)\frac{\mathbf u\mathbf u^T}{|\mathbf u|}$$

In a coordinate system aligned with the flow, the longitudinal and transverse dispersion coefficients are:

$$D_L=\alpha_L|\mathbf u|$$

$$D_T=\alpha_T|\mathbf u|$$

Including molecular diffusion:

$$D_{L,\mathrm{total}}=\alpha_L|\mathbf u|+D_m^*$$

$$D_{T,\mathrm{total}}=\alpha_T|\mathbf u|+D_m^*$$

The corresponding dispersive mass flux is:

$$\mathbf j_d=-\phi_e\mathbf D\nabla c$$
